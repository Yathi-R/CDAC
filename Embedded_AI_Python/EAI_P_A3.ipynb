```python
import pickle

class Product:
    def read(self):
        self.ID = input("Enter Product ID: ")
        self.price = float(input("Enter Product Price: "))

    def display(self):
        print(f"Product ID: {self.ID}")
        print(f"Price: {self.price}")

    def discounted_price(self):
        if self.price > 10000:
            discount = self.price * 0.05
        else:
            discount = 0
        return self.price - discount


product_list = []

for _ in range(3):
    p = Product()
    p.read()
    p.display()
    discount_price = p.discounted_price()
    print(f"Discounted Price: {discount_price}")
    product_list.append([p.ID, p.price, discount_price])

file = open("products.dat", "wb")
pickle.dump(product_list, file)
file.close()

file = open("products.dat", "rb")
loaded_list = pickle.load(file)
file.close()

print("\nProducts read from file:")
for item in loaded_list:
    ID, price, discount_price = item
    print(f"Product ID: {ID}")
    print(f"Price: {price}")
    print(f"Discounted Price: {discount_price}")
    print("-" * 20)
```
