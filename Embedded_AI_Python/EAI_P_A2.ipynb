```python
#1)Count Vowels in a String

def count_vowels(text):
    vowels = "aeiou"
    count = 0

    text = text.lower()

    for char in text:
        if char in vowels:
            count += 1

    return count

str1 = input("Enter a string: ")
print("Number of vowels:", count_vowels(str1))
```

```python
#2) Check Palindrome

def is_palindrome(text):
    n = len(text)
    
    for i in range(n//2):
        if text[i] == text[-i -1]:
            bol = True
        else:
            bol = False
    if bol == True:
        print(f"{text} is a palindrome")
    else:
        print(f"{text} is not a palindrome")
        
str2 = input("Enter a string: ")
is_palindrome(str2)
```

```python
#4) Sum of All Elements in a List

def sum_list(ls):
    sum = 0

    for ele in ls:
        sum += ele

    print(sum)


ls = [10, 20, 30, 40, 50]
sum_list(ls)
```

```python
#5) Find Maximum and Minimum in a List

def max_list(ls1):
    l_max = 0
    for ele in ls1:
        if l_max < ele:
            l_max = ele
    print("Max in the list:", l_max)

def min_list(ls1):
    l_min = 0
    for ele in ls1:
        if l_min > ele:
            l_min = ele
    print("Min in the list:", l_min)

ls1 = [15, -67, 94, 26, -29]
max_list(ls1)
min_list(ls1)
```

```python
#6) Remove Duplicates from List

def remove_duplicates(ls3):
    new_list = []
    for item in ls3:
        if item not in new_list:
            new_list.append(item)
    return new_list

ls3 = [1, 2, 3, 2, 1, 6, 5, 6, 5, 9, 0, 1, 9]
new_list = remove_duplicates(ls3)
print(new_list)
```

```python
#7)Count Occurrences of an Element in Tuple

def count_ele(tp1, n):
    count = tp1.count(n)
    return count

    
tp1 = (1, 2, 3, 4, 2, 3, 1, 4, 5, 3, 2)
count = count_ele(tp1, 2)
print(count)
```

```python
#8)Find Student with Highest Marks

def top_student(marks):
    highest = 0
    name = ""
    for student in marks:
        if marks[student] > highest:
            highest = marks[student]
            name = student
    return name

students = {
    "Soniya": 78,
    "Yathi": 92,
    "Shivansh": 85,
    "Ritish": 80
}

print("Student name with highest mark :", top_student(students))
```

```python
#9. Merging two dictionary
def merge_dicts(dict1, dict2):
    merged = {}

    for key in dict1:
        if key in dict2:
            merged[key] = [dict1[key], dict2[key]]
        else:
            merged[key] = dict1[key]

    for key in dict2:
        if key not in dict1:
            merged[key] = dict2[key]
    
    return merged

dict1 = {"x": 1, "y": 2, "z": 3}
dict2 = {"a": 4, "b": 5, "x": 6}

result = merge_dicts(dict1, dict2)
print(result)
```

```python
#10. string as input and returns a dictionary with each char and count of char
def count_characters(text):
    char_count = {}
    for char in text:
        if char in char_count:
            char_count[char] += 1
        else:
            char_count[char] = 1
    return char_count

str5 = input("Enter a string ")
result = count_characters(str5)
print(result)
```
